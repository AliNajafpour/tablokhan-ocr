```python
normalizer = hazm.Normalizer()

for i in range(len(papers)):
    papers[i] = normalizer.normalize(papers[i])
```

```python
tokenized = []
for paper in papers:
    tokenized.append(hazm.word_tokenize(paper))
```

```python
tokenized_len = [len(a) for a in tokenized]
```

```python

plt.figure()
plt.hist(tokenized_len)
plt.ylabel("number of papers")
plt.xlabel("lenght")
plt.title("histogram of lengh of papers")

plt.show()
```

```python
tk_mean = np.mean(tokenized_len)
tk_std = np.std(tokenized_len)

tk_lower_bound = tk_mean - 2 * tk_std
tk_upper_bound = tk_mean + 2 * tk_std

print(tk_mean)
print(tk_std)
print(tk_lower_bound)
print(tk_upper_bound)
```

```python
# حذف کردن مقاله ها زیر 50 کلمه

lowerbound = 50

keep = [len(tokens) >= lower_bound for tokens in tokenized]

papers = [paper for paper, ok in zip(papers, keep) if ok]

```

```python
hazm.sent_tokenize(papers[0])
```

```python
sentences = []

for paper in papers:
    normalized_paper = normalizer.normalize(paper)

    for sentence in hazm.sent_tokenize(normalized_paper):
        sentence = sentence.strip()

        if sentence:
            sentences.append(sentence)
```

```python
useless = {
    ".", ",", "،", "؛", ":", "!", "?", "؟",
    "(", ")", "[", "]", "{", "}", "«", "»", "…"
}


tokenized_sent = []

for sentence in sentences:
    tokenized_sent.append(hazm.word_tokenize(sentence))

for sentence in tokenized_sent:
    for word in sentence:
        if word in useless:
            sentence.remove(word)

print(tokenized_sent[:10])


```

```python
three_grams = []

for t in tokenized_sent:
    for i in range(len(t) - 2):
        three_grams.append(" ".join(t[i:i + 3]))

print(three_grams[:10])
```

```python
len(three_grams)
```

```python
unique_grams = set(three_grams)

frequency = dict()

for gram in unique_grams:
    frequency[gram] = three_grams.count(gram)


frequency_values = frequency.values()

frequency_mean = np.mean(frequency_values)
frequency_std = np.std(frequency_values)

frequency_lower_bound = frequency_mean - 2 * frequency_std
frequency_upper_bound = frequency_mean + 2 * frequency_std

print(frequency_mean)
print(frequency_std)

# زیادی طول کشید این اجرا نکنید این روش رو
```

```python
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer


vectorizer = TfidfVectorizer(
    tokenizer=str.split,
    token_pattern=None,
    lowercase=False,
    dtype=np.float32
)

tfidf = vectorizer.fit_transform(unique_grams)

print("Matrix shape:", tfidf.shape)
print("Stored values:", tfidf.nnz)
```

```python
from sklearn.cluster import MiniBatchKMeans

kmeans = MiniBatchKMeans(n_clusters=100,batch_size=4096,n_init=1,random_state=42,)

labels = kmeans.fit_predict(tfidf)

# اجرا نشود  غیر ممکن به خاطر تایمی که میبرد
```

```python
from sklearn.decomposition import TruncatedSVD
from sklearn.preprocessing import normalize
from sklearn.cluster import MiniBatchKMeans

tfidf = tfidf.astype("float32")

tfidf_reduced = TruncatedSVD(
    n_components=128,
    random_state=42
).fit_transform(tfidf)

tfidf_reduced = normalize(tfidf_reduced)

kmeans = MiniBatchKMeans(
    n_clusters=1000,
    batch_size=4096,
    n_init=1,
    random_state=42
)

labels = kmeans.fit_predict(tfidf_reduced)
```

```python
print(labels[:10])
```

```python
len(np.unique(labels))
```

```python
selected = []

for c in np.unique(labels):
    ids = np.where(labels == c)[0]
    distances = ((tfidf_reduced[ids] - kmeans.cluster_centers_[c]) ** 2).sum(1)
    selected.append(ids[distances.argmin()])

selected_grams = [list(unique_grams)[i] for i in selected]
```

```python
print("TF-IDF rows:", tfidf.shape[0])
print("Total labels:", labels.shape[0])
print("Used clusters:", np.unique(labels).size)
```
